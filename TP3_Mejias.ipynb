```python
# Ejercicio 1: Slicing y Sub-matrices
import numpy as np

matriz = np.arange(1, 26).reshape(5, 5)
print("Matriz 5x5:")
print(matriz)

submatriz_central = matriz[1:4, 1:4]
print("\nSub-matriz central 3x3:")
print(submatriz_central)

ultima_columna = matriz[:, -1]
print("\nÚltima columna:")
print(ultima_columna)
```

```python
# Ejercicio 2: Estadísticas con Ejes (Axis)
ventas = np.array([
    [1500, 2000, 1800, 2200],
    [1100, 1300, 1200, 1500],
    [3000, 3100, 2900, 3500]
])

print("Venta total de toda la empresa:", np.sum(ventas))
print("Promedio de ventas por semana (axis=0):", np.mean(ventas, axis=0))
print("Venta máxima por sucursal (axis=1):", np.max(ventas, axis=1))
```

```python
# Ejercicio 3: El poder de la Máscara Booleana
numeros = np.arange(1, 21)

mascara_pares = numeros % 2 == 0

pares = numeros[mascara_pares]
print("Números pares:", pares)
```

```python
# Ejercicio 4: Condiciones Múltiples (El Analista de RRHH)
edades = np.array([22, 55, 34, 19, 45, 61, 28, 30, 42])

entre_30_y_50 = edades[(edades >= 30) & (edades <= 50)]
print("Empleados entre 30 y 50 años:", entre_30_y_50)

extremos = edades[(edades < 25) | (edades > 55)]
print("Empleados menores de 25 o mayores de 55:", extremos)
```

```python
# Ejercicio 5: Modificación Condicional
sensor = np.zeros((4, 4))

sensor[1:3, 1:3] = 9

print("Matriz sensor con bloque central calentado:")
print(sensor)
```

```python
# Ejercicio 6: Desafío Integrador - El Profesor Robot
calificaciones = np.random.randint(1, 11, 50)
print("Calificaciones generadas:", calificaciones)

promedio = np.mean(calificaciones)
print(f"\nPromedio general del curso: {promedio:.2f}")

desaprobados = calificaciones[calificaciones < 4]
print(f"Cantidad de alumnos desaprobados: {desaprobados.size}")

calificaciones[calificaciones < 4] = 4

print("\nCalificaciones finales (sin aplazos):")
print(calificaciones)
```
