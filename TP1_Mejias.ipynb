```python
# Ejercicio 2: Operaciones con Diccionarios
producto = {
    "id_articulo": 1054,
    "nombre": "Monitor 24 pulgadas",
    "precio": 150000,
    "stock": 15
}

producto["stock"] -= 1
producto["categoria"] = "Tecnología"

print(f"Se vendió un {producto['nombre']}. Quedan {producto['stock']} en stock.")
```

```python
# Ejercicio 3: Limpieza rápida con List Comprehension
edades_crudas = [21, "25", -5, 30, "19", 45, -1]

edades_validas = [int(e) for e in edades_crudas if int(e) > 0]

print(edades_validas)
```

```python
# Ejercicio 4: El Mini-ETL Financiero
ventas = [
    {"ticket": 1, "monto": 5000, "metodo": "Efectivo"},
    {"ticket": 2, "monto": 12000, "metodo": "Tarjeta"},
    {"ticket": 3, "monto": 8500, "metodo": "Efectivo"},
    {"ticket": 4, "monto": 3000, "metodo": "MercadoPago"}
]

total_dia = sum(v["monto"] for v in ventas)
total_efectivo = sum(v["monto"] for v in ventas if v["metodo"] == "Efectivo")

print(f"Monto total recaudado en el día: ${total_dia}")
print(f"Monto total recaudado en Efectivo: ${total_efectivo}")
```
