```python
# Ejercicio 1: Creación y Atributos
import numpy as np

edades = np.array([22, 25, 30, 35, 40])

print("Dimensiones:", edades.ndim)
print("Forma:", edades.shape)
print("Tipo de dato:", edades.dtype)

edades_float = edades.astype(np.float64)
print("Array en float64:", edades_float)
```

```python
# Ejercicio 2: Secuencias Automáticas
ids_clientes = np.arange(1, 101)
puntos_grafico = np.linspace(0, 10, 50)

print("IDs clientes:", ids_clientes)
print("Puntos gráfico:", puntos_grafico)
```

```python
# Ejercicio 3: Matrices y Datos Aleatorios
matriz_aleatoria = np.random.rand(3, 4)
matriz_unos = np.ones((3, 4))

print("Matriz aleatoria 3x4:")
print(matriz_aleatoria)
print("\nMatriz de unos 3x4:")
print(matriz_unos)
```

```python
# Ejercicio 4: El Simulador de Inventario
sucursales = int(input("Ingrese la cantidad de Sucursales: "))
productos = int(input("Ingrese la cantidad de Productos: "))

inventario = np.zeros((sucursales, productos))

print("Base de datos de inventario inicializada correctamente")
print(inventario)
```

```python
# Ejercicio 5: Redimensionando la realidad (Reshape)
vector = np.arange(1, 13)
matriz = vector.reshape(3, 4)

print("Matriz 3x4:")
print(matriz)
print("Dimensiones:", matriz.ndim)
```

```python
# Ejercicio 6: El peso de los datos (Optimización)
temperaturas = np.array([25.5, 28.1, 22.4, 19.9, 30.2])
print("Tipo original:", temperaturas.dtype)

temperaturas_int = temperaturas.astype(np.int32)
print("Array convertido a int32:", temperaturas_int)
```

```python
# Ejercicio 7: Simulador de Dados
tiradas = np.random.randint(1, 7, 10)
print("10 tiradas de dado:", tiradas)
```

```python
# Ejercicio 8: Limpieza de sensores
humedad_texto = np.array(["45.2", "48.1", "42.5", "50.0"])
print("Array como texto:", humedad_texto)
print("Tipo de dato:", humedad_texto.dtype)

try:
    np.sum(humedad_texto)
except TypeError:
    print("Error: el array tiene dtype <U4 (texto Unicode). NumPy no puede realizar operaciones aritméticas sobre strings. Es necesario convertirlo a float64 con .astype() antes de operar.")

humedad_float = humedad_texto.astype(np.float64)
print("Array limpio float64:", humedad_float)
```

```python
# Ejercicio 9: La Matriz Identidad
identidad = np.eye(4)
print("Matriz identidad 4x4:")
print(identidad)

print("\nMatriz identidad x 5:")
print(identidad * 5)
```
