```python
# Ejercicio 1: Creación del entorno de datos
import csv

datos_tienda = [
    {"ID": 1001, "Producto": "Laptop Gamer", "Categoria": "Computacion", "Precio": 1200000, "Stock": 15},
    {"ID": 1002, "Producto": "Mouse Inalambrico", "Categoria": "Accesorios", "Precio": 25000, "Stock": 100},
    {"ID": 1003, "Producto": "Teclado Mecanico", "Categoria": "Accesorios", "Precio": 85000, "Stock": 40},
    {"ID": 1004, "Producto": "Monitor 27 Pulgadas", "Categoria": "Computacion", "Precio": 350000, "Stock": 25},
    {"ID": 1005, "Producto": "Auriculares Bluetooth", "Categoria": "Audio", "Precio": 60000, "Stock": ""},
    {"ID": 1006, "Producto": "Silla Ergonomica", "Categoria": "Muebles", "Precio": 250000, "Stock": 10}
]

with open('productos.csv', 'w', newline='', encoding='utf-8') as archivo:
    columnas = ["ID", "Producto", "Categoria", "Precio", "Stock"]
    writer = csv.DictWriter(archivo, fieldnames=columnas)
    writer.writeheader()
    writer.writerows(datos_tienda)

print("Archivo 'productos.csv' creado exitosamente.")
```

```python
# Ejercicio 2: Ingesta y el "Vistazo" inicial
import pandas as pd

df_productos = pd.read_csv('productos.csv')

print("Primeras 3 filas:")
print(df_productos.head(3))

print("\nÚltimas 2 filas:")
print(df_productos.tail(2))
```

```python
# Ejercicio 3: La Auditoría Técnica y Estadística

# Auditoría de calidad
print("=== Resumen técnico ===")
print(df_productos.info())
print("\nConclusión: La columna 'Stock' tiene 5 valores no nulos sobre 6 registros. Hay 1 dato faltante (Auriculares Bluetooth).")

# Auditoría financiera
print("\n=== Resumen estadístico ===")
print(df_productos.describe())
print(f"\nConclusión: El precio promedio de los productos es ${df_productos['Precio'].mean():,.0f}. El producto más barato cuesta ${df_productos['Precio'].min():,.0f}.")
```

```python
# Ejercicio 4: Extracción de Series vs. DataFrames

# DataFrame con doble corchete
tabla_marketing = df_productos[["Producto", "Precio"]]
print("tabla_marketing (DataFrame):")
print(tabla_marketing)

# Serie con un solo corchete
serie_precios = df_productos["Precio"]
print("\nSerie de precios:")
print(serie_precios)
print("\nTipo de dato:", type(serie_precios))
```

```python
# Ejercicio 5: El Desafío del Analista (Operaciones Vectorizadas)

# Convertimos Stock a numérico: Pandas transforma el valor vacío en NaN automáticamente
df_productos["Stock"] = pd.to_numeric(df_productos["Stock"], errors='coerce')

capital_por_producto = df_productos["Precio"] * df_productos["Stock"]

print("Capital potencial por producto:")
print(capital_por_producto)
print("\nNota: El producto con stock vacío (Auriculares Bluetooth) muestra NaN porque Pandas no puede operar con un valor nulo.")
```
