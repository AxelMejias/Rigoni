```python
# Ejercicio 1 - Parte 1: Generación del archivo CSV
import pandas as pd
import numpy as np

np.random.seed(123)
legajos_generados = ['TUP-' + str(i) for i in range(1001, 1021)]
carreras_disponibles = ['Programacion', 'Redes', 'Sistemas']

datos_academicos = {
    'Legajo': legajos_generados,
    'Carrera': np.random.choice(carreras_disponibles, 20),
    'Matematica': np.random.randint(2, 11, 20),
    'Bases_Datos': np.random.randint(2, 11, 20),
    'Algoritmos': np.random.randint(2, 11, 20)
}

df_generador = pd.DataFrame(datos_academicos)
df_generador.to_csv('notas_semestre.csv', index=False)
print('Archivo notas_semestre.csv generado con éxito.')
```

```python
# Ejercicio 1 - Parte 2: Lectura, índice y vistazo
df_notas = pd.read_csv('notas_semestre.csv')

df_notas = df_notas.set_index('Legajo')

print("Primeras 5 filas con Legajo como índice:")
print(df_notas.head(5))
```

```python
# Ejercicio 2: Auditoría y Navegación Ciega (iloc)

print("=== Auditoría estadística ===")
print(df_notas.describe())

print("\n=== Décima fila (índice 9) ===")
print(df_notas.iloc[9])

print("\n=== Primeras 3 filas y primeras 2 columnas ===")
print(df_notas.iloc[:3, :2])
```

```python
# Ejercicio 3: Operando al Paciente (loc)

df_notas.loc['TUP-1005', 'Bases_Datos'] = 9
df_notas.loc['TUP-1018', 'Matematica'] = 7

print("Verificación de modificaciones:")
print(df_notas.loc[['TUP-1005', 'TUP-1018']])
```

```python
# Ejercicio 4: Integración - Máscaras Booleanas Múltiples

mascara = (df_notas['Carrera'] == 'Programacion') & (df_notas['Algoritmos'] >= 8)

df_destacados = df_notas[mascara]

print("Alumnos destacados de Programacion con Algoritmos >= 8:")
print(df_destacados)
```

```python
# Ejercicio 5: El Desafío Final (Estadística Vectorizada)

df_redes = df_notas[df_notas['Carrera'] == 'Redes']

promedio_matematica_redes = df_redes['Matematica'].mean()

print(f"Promedio de Matemática de los alumnos de Redes: {promedio_matematica_redes:.2f}")
```
